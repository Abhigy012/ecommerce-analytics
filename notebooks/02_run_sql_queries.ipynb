# 02 · Run SQL analysis queries

Runs every `.sql` file in `sql/queries/` against `db/olist.db` and saves
the result as a CSV in `results/sql_outputs/`.

```python
import sqlite3
import pandas as pd
from pathlib import Path

ROOT = Path("..").resolve()
DB_PATH = ROOT / "db" / "olist.db"
QUERIES_DIR = ROOT / "sql" / "queries"
OUT_DIR = ROOT / "results" / "sql_outputs"
OUT_DIR.mkdir(parents=True, exist_ok=True)
```

```python
conn = sqlite3.connect(DB_PATH)
for sql_file in sorted(QUERIES_DIR.glob("*.sql")):
    query = sql_file.read_text()
    df = pd.read_sql_query(query, conn)
    out_path = OUT_DIR / f"{sql_file.stem}.csv"
    df.to_csv(out_path, index=False)
    print(f"\n=== {sql_file.name} -> {out_path.relative_to(ROOT)} ({len(df)} rows) ===")
    print(df.head(10).to_string(index=False))
conn.close()
```
